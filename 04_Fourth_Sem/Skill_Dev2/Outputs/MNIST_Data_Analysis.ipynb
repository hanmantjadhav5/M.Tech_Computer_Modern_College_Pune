```python
# Library Import
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Dataset Load
data = pd.read_csv("C:/Users/hanmant/Desktop/sdAssignment/Assignment7/mnist_train.csv")
data.head()
```

```python
# Dataset Information
data.info()
```

```python
# Statistical Summary
data.describe()
```

```python
# Distribution of Digit Labels (Visualization)
data['label'].value_counts().sort_index().plot(kind='bar')
plt.title("Distribution of Handwritten Digit Labels (MNIST)")
plt.xlabel("Digit")
plt.ylabel("Count")
plt.show()
```

```python
# digit image visualize
image = data.iloc[0,1:].values
image = image.reshape(28,28)
plt.imshow(image, cmap='gray')
plt.title(f"Label : {data.iloc[0,0]}")
plt.axis("off")
plt.show()
```
