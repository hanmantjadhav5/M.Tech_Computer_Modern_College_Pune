```python
# Load MNIST CSV using Pandas
import pandas as pd

data = pd.read_csv("/content/MNIST_Data/mnist_train.csv")

data.head()
```

```python
# Import NumPy
import numpy as np
```

```python
# Mean pixel intensity (excluding label column)

mean_pixel = data.iloc[:, 1:].mean().mean()
mean_pixel
```

```python
# Variance of pixel intensities

variance_pixel = data.iloc[:, 1:].var().mean()
variance_pixel
```

```python
# Standard deviation

std_pixel = data.iloc[:, 1:].std().mean()
std_pixel
```

```python
# Label Count (0–9 digits distribution)
label_counts = data['label'].value_counts().sort_index()
label_counts
```

```python
# Label Statistics
label_counts.describe()
```

```python
# Histogram of Pixel Intensities
import matplotlib.pyplot as plt

pixels = data.iloc[:,1:].values.flatten()

plt.hist(pixels, bins=50)
plt.title("Histogram of Pixel Intensities")
plt.xlabel("Pixel Intensity")
plt.ylabel("Frequency")
plt.show()
```
